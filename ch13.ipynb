## 13.1 질문과 데이터

````python
# C01 | 점검 함수: 행·열 수와 형식·결측·기간·중복을 한 번에 봅니다.
import pandas as pd


def first_check(df, time_col):
    print(len(df), '행 ×', len(df.columns), '열')
    print('열 형식:', df.dtypes.astype(str).to_dict())
    print('결측:', df.isna().sum().to_dict())
    print('기간:', df[time_col].min(), '~', df[time_col].max())
    print('중복 행:', df.duplicated().sum())
````

````python
# C02 | 점검 연습: 문제를 일부러 섞은 작은 표로 함수를 시험합니다.
sample = pd.DataFrame({
    'month': [1, 2, 2, 3, 4, 5],
    'region': ['서울', '서울', '서울', '부산', '부산', None],
    'value': [120, 135, 135, None, 98, 101]})
first_check(sample, 'month')
````

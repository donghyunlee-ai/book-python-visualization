## 11.2 데이터·위젯·필터

```python
# C01 | 앱 자료: 12개월 매출·영업이익 표를 앱 폴더에 CSV로 저장합니다.
from pathlib import Path
import pandas as pd

sales_a = [100, 120, 150, 130, 140, 160, 150, 170, 180, 160, 190, 200]
sales_b = [70, 90, 100, 120, 110, 130, 140, 150, 160, 170, 160, 180]
profit_a = [10, 15, 25, 20, 21, 25, 22, 28, 29, 26, 32, 35]
profit_b = [7, 9, 12, 18, 15, 17, 18, 21, 23, 24, 22, 27]
data = pd.DataFrame({'month': list(range(1, 13)) * 2,
                     'branch': ['A'] * 12 + ['B'] * 12,
                     'sales': sales_a + sales_b,
                     'profit': profit_a + profit_b})
app_dir = Path('ch11_app')
app_dir.mkdir(exist_ok=True)
data.to_csv(app_dir / 'sales.csv', index=False)   # 행 라벨은 빼고 저장
print(len(data), '행')
print(data.groupby('branch')[['sales', 'profit']].sum())
```

## 11.4 검산과 오류 진단

```python
# C03 | 조건별 검산: 앱에서 고를 조건마다 기대하는 숫자를 계산합니다.
cases = [(['A', 'B'], (1, 12), 'sales'), (['A'], (1, 4), 'sales'),
         (['A'], (1, 4), 'profit'), (['B'], (10, 10), 'profit'),
         ([], (1, 12), 'sales')]
for branches, months, metric in cases:
    part = data[data['branch'].isin(branches)
                & data['month'].between(months[0], months[1])]
    if part.empty:
        print(branches, months, metric, '→ 경고 후 멈춤')
    else:
        print(branches, months, metric, '→',
              f'{len(part)}행, {part[metric].sum():,}')
```

```python
# C04 | 기간 영업이익률: 지점의 기간 영업이익률을 두 방식으로 구합니다.
period = data.groupby('branch')[['sales', 'profit']].sum()
period['margin'] = period['profit'] / period['sales'] * 100
monthly_margin = data['profit'] / data['sales'] * 100
print(period['margin'].round(1).to_dict())
print(monthly_margin.groupby(data['branch']).mean().round(1).to_dict())
```

### W11-01

```python
# W11-01 | 풀이: A02의 default를 A지점만으로 바꾸고 첫 화면을 검산합니다.
# 1. A02의 한 줄을 이렇게 바꾸고 저장한 뒤, 브라우저에서 Rerun을 누릅니다.
#    branches = st.sidebar.multiselect('지점', ['A', 'B'], default=['A'])
# 아래 셀은 앱 파일을 직접 고치는 대신, 고친 사본을 만들어 AppTest(부록 C)로 첫 화면을 읽는 확인 코드입니다.
# try_app()은 W11-02~W11-04에서도 씁니다.
from pathlib import Path
from streamlit.testing.v1 import AppTest


def try_app(changes, name):
    # ch11_app/app.py에서 (원래 줄, 고친 줄)을 바꾼 사본 name을 만들어 실행합니다.
    # 원본은 그대로이고, 사본(app_w11_0N.py)은 ch11_app에 남으니 확인 뒤 지워도 됩니다.
    source = Path('ch11_app/app.py').read_text(encoding='utf-8')
    for old, new in changes:
        assert old in source, old
        source = source.replace(old, new)
    variant = Path('ch11_app') / name
    variant.write_text(source, encoding='utf-8')
    return AppTest.from_file(str(variant.resolve()), default_timeout=30).run()


at = try_app([("default=['A', 'B'])", "default=['A'])")], 'app_w11_01.py')
print('2. 첫 화면의 카드:', [m.value for m in at.metric])
```

```python
# 3. C02를 새 셀에 복사해 첫 화면 조건으로
branches, months, metric = ['A'], (1, 12), 'sales'
view = data[data['branch'].isin(branches)
            & data['month'].between(months[0], months[1])]
print(len(view), '행 →', view[metric].sum(), '백만원')
assert [m.value for m in at.metric] == [f'{view[metric].sum():,}백만원', f'{len(view)}행']
```

### W11-02

```python
# W11-02 | 풀이: 세 번째 카드로 한 행 평균을 더합니다.
# 1·2. A03에서 두 곳을 바꿉니다.
#    c1, c2, c3 = st.columns(3)
#    c3.metric('한 행 평균', f'{view[metric].mean():,.1f}백만원')   # c2.metric() 줄 아래에
changes = [("c1, c2 = st.columns(2)\n", "c1, c2, c3 = st.columns(3)\n"),
           ("c2.metric('관측 수', f'{len(view)}행')\n",
            "c2.metric('관측 수', f'{len(view)}행')\n"
            "c3.metric('한 행 평균', f'{view[metric].mean():,.1f}백만원')\n")]
at = try_app(changes, 'app_w11_02.py')
at.multiselect[0].set_value(['A']).run()
at.slider[0].set_value((1, 4)).run()
print('매출 카드:', [m.value for m in at.metric])
at.selectbox[0].set_value('영업이익').run()
print('영업이익 카드:', [m.value for m in at.metric])
```

```python
# 3. 노트북에서 같은 조건으로 계산해 대조
for metric, label in [('sales', '매출'), ('profit', '영업이익')]:
    view = data[data['branch'].isin(['A']) & data['month'].between(1, 4)]
    print(label, '→', view[metric].sum(), len(view), round(view[metric].mean(), 1))
assert [m.value for m in at.metric] == ['70백만원', '4행', '17.5백만원']
```

### W11-03

```python
# W11-03 | 풀이: 합계 막대가 view 대신 data를 집계하게 일부러 끊고, 차이를 확인합니다.
# 1. A03의 한 줄을 이렇게 바꿉니다(끊긴 연결).
#    totals = data.groupby('branch', as_index=False)[metric].sum()
at = try_app([("totals = view.groupby(", "totals = data.groupby(")], 'app_w11_03.py')
at.multiselect[0].set_value(['A']).run()
at.slider[0].set_value((1, 4)).run()
at.selectbox[0].set_value('영업이익').run()
print('2. 카드:', [m.value for m in at.metric])
```

```python
# 막대가 집계한 값과 카드가 집계한 값을 노트북에서 견줍니다.
view = data[data['branch'].isin(['A']) & data['month'].between(1, 4)]
print('카드(view):', view['profit'].sum(), '| 끊긴 막대(data):',
      data.groupby('branch')['profit'].sum().to_dict())
assert at.metric[0].value == '70백만원' and data.groupby('branch')['profit'].sum()['A'] == 288
# 3. view로 되돌리면 막대도 A 70이 된다.
# 답(예): 끊긴 막대는 고른 조건과 상관없이 전체 표 data(1~12월, 두 지점)를 집계해 A 288·B 213백만원을 그렸다.
```

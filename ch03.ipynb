## 3.1 단일 차트에서 다중 차트로

```python
# C01 | 공통 데이터: A지점과 B지점의 1~4월 실적입니다.
import matplotlib.pyplot as plt

months = ['1월', '2월', '3월', '4월']
sales = [100, 120, 150, 130]        # A지점 매출 (백만원)
profit = [10, 15, 25, 20]           # A지점 영업이익 (백만원)
sales_b = [70, 90, 100, 120]        # B지점 매출 (백만원)
marketing = [10, 12, 15, 13]        # A지점 광고비 (백만원)
conversion = [2.0, 2.4, 2.2, 2.8]   # A지점 전환율 (%)
```

```python
# C02 | 한 Axes에 두 선: 같은 ax에 plot()을 두 번 부릅니다.
fig, ax = plt.subplots(figsize=(6, 3.6))
ax.plot(months, sales, marker='o', color='#0072B2', label='매출')
ax.plot(months, profit, marker='^', linestyle=':', color='#0072B2',
        label='영업이익')
ax.set(title='A지점 1~4월 실적', xlabel='월',
       ylabel='금액 (백만원)', ylim=(0, 180))
ax.legend(loc='upper left')         # label에 적은 이름으로 범례 표시
fig.tight_layout()
plt.show()
```

## 3.2 Axes 배치와 접근

```python
# C03 | 1행 2열: 왼쪽은 axes[0], 오른쪽은 axes[1]입니다.
fig, axes = plt.subplots(1, 2, figsize=(6, 2.7))
axes[0].plot(months, sales, marker='o', color='#0072B2')
axes[0].set(title='매출', xlabel='월',
            ylabel='매출 (백만원)', ylim=(0, 160))
axes[1].plot(months, profit, marker='^', linestyle=':', color='#0072B2')
axes[1].set(title='영업이익', xlabel='월',
            ylabel='영업이익 (백만원)', ylim=(0, 30))
fig.tight_layout()
plt.show()
```

```python
# C04 | 반환 모양: 배치에 따라 두 번째 반환값의 모양이 달라집니다.
fig1, ax1 = plt.subplots()          # 1×1
fig2, axes2 = plt.subplots(1, 2)    # 1×2
fig3, axes3 = plt.subplots(2, 1)    # 2×1
fig4, axes4 = plt.subplots(2, 2)    # 2×2

print('1×1:', type(ax1).__name__)
print('1×2:', type(axes2).__name__, axes2.shape)
print('2×1:', type(axes3).__name__, axes3.shape)
print('2×2:', type(axes4).__name__, axes4.shape)
print('2×2의 axes4[1]:', type(axes4[1]).__name__, axes4[1].shape)
print('2×2의 axes4[1, 0]:', type(axes4[1, 0]).__name__)
plt.close('all')                    # 확인용 빈 그림 네 개 닫기
```

```python
# C05 | 2행 2열: axes[행, 열]로 패널 하나를 고릅니다.
fig, axes = plt.subplots(2, 2, figsize=(6, 2.8))
axes[0, 0].set_title('axes[0, 0]')
axes[0, 1].set_title('axes[0, 1]')
axes[1, 0].set_title('axes[1, 0]')
axes[1, 1].set_title('axes[1, 1]')
fig.tight_layout()
plt.show()
```

```python
# C06 | 기본 차트 여섯 가지: 패널마다 다른 그리기 메서드를 부릅니다.
fig, axes = plt.subplots(2, 3, figsize=(6, 3.0))
axes[0, 0].bar(months, sales)                         # 비교
axes[0, 0].set_title('bar() 막대')
axes[0, 1].barh(['A지점', 'B지점'], [sales[3], sales_b[3]])
axes[0, 1].set_title('barh() 가로 막대')
axes[0, 2].plot(months, sales, marker='o')            # 추세
axes[0, 2].set_title('plot() 선')
axes[1, 0].hist(sales + sales_b, bins=4)              # 분포
axes[1, 0].set_title('hist() 히스토그램')
axes[1, 1].scatter(marketing, sales)                  # 관계
axes[1, 1].set_title('scatter() 산점도')
axes[1, 2].pie([sales[3], sales_b[3]], labels=['A지점', 'B지점'],
               startangle=90)                         # 비중
axes[1, 2].set_title('pie() 원')
fig.tight_layout()
plt.show()
```

## 3.3 축 공유와 공정한 비교

```python
# C07 | 자동 범위: 각 Axes가 자기 데이터에 맞춰 Y축 범위를 정합니다.
fig, axes = plt.subplots(1, 2, figsize=(6, 2.7))
axes[0].plot(months, sales, marker='o', color='#0072B2')
axes[0].set(title='A지점', xlabel='월', ylabel='매출 (백만원)')
axes[1].plot(months, sales_b, marker='s', linestyle='--', color='#D55E00')
axes[1].set(title='B지점', xlabel='월', ylabel='매출 (백만원)')
fig.tight_layout()
plt.show()

a_min, a_max = axes[0].get_ylim()
b_min, b_max = axes[1].get_ylim()
print('A지점 Y축:', a_min, '~', a_max)
print('B지점 Y축:', b_min, '~', b_max)
```

```python
# C08 | Y축 공유: 두 Axes가 Y축의 범위와 눈금을 함께 씁니다.
fig, axes = plt.subplots(1, 2, sharey=True, figsize=(6, 2.7))
axes[0].plot(months, sales, marker='o', color='#0072B2')
axes[0].set(title='A지점', xlabel='월', ylabel='매출 (백만원)')
axes[1].plot(months, sales_b, marker='s', linestyle='--', color='#D55E00')
axes[1].set(title='B지점', xlabel='월')
axes[0].set_ylim(0, 160)  # 공유된 axes[1]에도 같은 범위가 걸립니다
fig.tight_layout()
plt.show()

a_min, a_max = axes[0].get_ylim()
b_min, b_max = axes[1].get_ylim()
print('A지점 Y축:', a_min, '~', a_max)
print('B지점 Y축:', b_min, '~', b_max)
```

### 직접 해 보기

**W3-02** C08로 축 공유를 확인하고 두 지점의 차이를 계산해 해석합니다.

```python
# W3-02 | 풀이: C08의 범위를 오른쪽 Axes에 걸고, 오른쪽에도 눈금 라벨을 보입니다.
fig, axes = plt.subplots(1, 2, sharey=True, figsize=(6, 2.7))
axes[0].plot(months, sales, marker='o', color='#0072B2')
axes[0].set(title='A지점', xlabel='월', ylabel='매출 (백만원)')
axes[1].plot(months, sales_b, marker='s', linestyle='--', color='#D55E00')
axes[1].set(title='B지점', xlabel='월')
axes[1].set_ylim(0, 160)                      # 1. 오른쪽에 걸어도 공유된 왼쪽에 같이 걸린다
axes[1].tick_params(labelleft=True)           # 2. 오른쪽 Axes에도 왼쪽 눈금 라벨
fig.tight_layout()
plt.show()

print('A지점 Y축:', axes[0].get_ylim(), '| B지점 Y축:', axes[1].get_ylim())
assert axes[0].get_ylim() == axes[1].get_ylim() == (0, 160)
```

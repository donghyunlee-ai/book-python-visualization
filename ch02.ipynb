## 2.1 Matplotlib과 실습 준비

````python
# C01 | 공통 데이터: A지점의 1~4월 실적입니다. 단위는 모두 백만원입니다.
months = ['1월', '2월', '3월', '4월']
sales = [100, 120, 150, 130]   # 매출
profit = [10, 15, 25, 20]      # 영업이익
````

## 2.2 Figure·Axes·Axis의 구조

## 2.3 두 가지 차트 작성 방식

````python
# C02 | plt 방식: 현재 Axes에 차례로 명령합니다.
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 3.6))              # 새 그림을 만들어 현재 그림으로
plt.plot(months, sales, marker='o', color='#0072B2')
plt.title('A지점 1~4월 매출 추이')      # 현재 Axes의 제목
plt.xlabel('월')
plt.ylabel('매출 (백만원)')
plt.tight_layout()
plt.show()
````

````python
# C03 | fig, ax 방식: 명령할 객체를 변수로 지정합니다.
fig, ax = plt.subplots(figsize=(6, 3.6))  # Figure와 Axes를 함께 만들기
ax.plot(months, sales, marker='o', color='#0072B2')
ax.set_title('A지점 1~4월 매출 추이')
ax.set_xlabel('월')
ax.set_ylabel('매출 (백만원)')
fig.tight_layout()                      # 여백 정리는 그림 전체의 일
plt.show()
````

## 2.4 단일 차트 완성과 저장

````python
# C04 | ax.set(): 제목·축 라벨·범위를 한 번에 설정합니다.
fig, ax = plt.subplots(figsize=(6, 3.6))
ax.plot(months, sales, marker='o', color='#0072B2')
ax.set(
    title='A지점 1~4월 매출 추이',
    xlabel='월',
    ylabel='매출 (백만원)',
    ylim=(0, 180),                      # 보이는 범위만 0~180으로
)
ax.grid(axis='y', alpha=0.25)           # 옅은 가로 보조선
fig.tight_layout()
plt.show()

ymin, ymax = ax.get_ylim()             # 지금 보이는 Y축 범위
print('표시 범위:', ymin, '~', ymax)
print('원래 데이터:', sales)
````

````python
# C05 | PNG 저장: 그림 전체(Figure)를 파일로 남깁니다.
from pathlib import Path

output_dir = Path('ch02_output')
output_dir.mkdir(exist_ok=True)         # 폴더가 이미 있어도 괜찮습니다
chart_path = output_dir / 'ch02_chart.png'

fig, ax = plt.subplots(figsize=(6, 3.6))
ax.plot(months, sales, marker='o', linewidth=2, color='#0072B2')
ax.set(title='A지점 1~4월 매출 추이', xlabel='월',
       ylabel='매출 (백만원)', ylim=(0, 180))
ax.grid(axis='y', alpha=0.25)
fig.tight_layout()
fig.savefig(chart_path, dpi=150, bbox_inches='tight')
plt.show()
````

````python
# C06 | 다시 열기: 저장한 파일을 열어 눈으로 확인합니다.
from IPython.display import Image, display

display(Image(filename=str(chart_path), width=600))
````

### W2-02

````python
# W2-02 | 풀이: 네 명령의 소유 객체와 하는 일입니다.
# 1. fig.savefig('chart.png') — Figure: 그림 전체를 파일로 저장한다.
# 2. ax.plot(months, sales)   — Axes: 그 차트 영역에 선을 그린다.
# 3. ax.set_xlim(...)         — Axes: X축의 보이는 범위를 정한다.
# 4. ax.xaxis                 — Axes의 속성: 괄호 없이 꺼내면 X축 객체(Axis 종류)를 돌려준다.
from matplotlib.axis import Axis

fig, ax = plt.subplots(figsize=(3, 2))
print('fig:', type(fig).__name__, '| ax:', type(ax).__name__,
      '| ax.xaxis:', type(ax.xaxis).__name__, '(Axis인가?', isinstance(ax.xaxis, Axis), ')')
plt.close(fig)
````

### W2-04

````python
# W2-04 | 풀이: 실패하는 이유와 가장 작은 수정입니다.
# 1. ax = plt.subplots()로 두 값(Figure, Axes)을 한 변수에 받아 ax가 튜플이다 → fig, ax = plt.subplots()
# 2. fig.plot(...): 선은 Figure가 아니라 Axes가 그린다           → ax.plot(months, sales)
# 3. ax.xlabel('월'): Axes의 메서드 이름은 set_xlabel이다          → ax.set_xlabel('월')
# 4. ax.yaxis.set_ylim(0, 180): 범위는 Axis가 아니라 Axes가 정한다 → ax.set_ylim(0, 180)
wrong = plt.subplots(figsize=(3, 2))         # 1번처럼 한 변수로 받으면 튜플
fig, ax = plt.subplots(figsize=(3, 2))        # 2~4번은 제대로 만든 fig, ax
checks = [('1', lambda: wrong.plot), ('2', lambda: fig.plot),
          ('3', lambda: ax.xlabel), ('4', lambda: ax.yaxis.set_ylim)]
for number, attr in checks:
    try:
        attr()
    except AttributeError as error:
        print(number, '→', error)
plt.close('all')
````

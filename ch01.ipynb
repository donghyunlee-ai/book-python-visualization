## 1.4 첫 차트 그리기

```python
# C01 | 첫 차트: A지점의 1~4월 매출을 선 그래프로 그립니다.
import matplotlib.pyplot as plt

months = ['1월', '2월', '3월', '4월']
sales = [100, 120, 150, 130]              # 매출, 단위: 백만원

fig, ax = plt.subplots(figsize=(6, 3.6))  # 그림 전체(fig)와 차트 영역(ax)
ax.plot(months, sales, marker='o')        # 월별 매출을 선으로 잇기
ax.set_title('A지점 1~4월 매출')
ax.set_xlabel('월')
ax.set_ylabel('매출 (백만원)')
plt.show()
```

## 1.5 AI 초안 점검하기

```python
# C02 | AI 초안: 검수 연습을 위해 만든 초안입니다.
fig, ax = plt.subplots(figsize=(6, 3.6))
ax.bar(months, sales, color=['#E69F00', '#56B4E9', '#009E73', '#CC79A7'])
ax.set_ylim(90, 155)
ax.set_title('매출 급성장!')
plt.show()
```

```python
# C03 | 초안 고치기: Y축을 0부터 두고 기간·단위·출처를 밝힙니다.
fig, ax = plt.subplots(figsize=(6, 3.6))
ax.bar(months, sales, color='#0072B2')
ax.set_ylim(0, 170)
ax.set_title('A지점 1~4월 매출')
ax.set_xlabel('월')
ax.set_ylabel('매출 (백만원)')
fig.text(0.99, 0.01, '자료: 교육용 가상 자료(월 합계)',
         ha='right', fontsize=9)
plt.show()
```

Pandas와 Seaborn을 활용하여 스타트업 매출 데이터를 분석해 봅니다.


```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 한글 폰트 설정 (Colab 기준)
plt.rc("font", family="NanumBarunGothic")
plt.rcParams["axes.unicode_minus"] = False
```

```
# 데이터 불러오기
df = pd.read_csv("data/tech_startup_sales.csv")
df["날짜"] = pd.to_datetime(df["날짜"])
df.head()
```

## 선 그래프 (시간의 흐름과 추이)


```
monthly_sales = df.groupby([df["날짜"].dt.to_period("M"), "부서"])["매출액"].sum().reset_index()
monthly_sales["날짜"] = monthly_sales["날짜"].astype(str)

plt.figure(figsize=(10, 5))
sns.lineplot(data=monthly_sales, x="날짜", y="매출액", hue="부서", marker="o")
plt.title("부서별 월간 매출 추이")
plt.xticks(rotation=45)
plt.show()
```

## 막대 그래프 (비교)


```
dept_sales = df.groupby("부서")["매출액"].sum().reset_index()
plt.figure(figsize=(8, 5))
sns.barplot(data=dept_sales, x="부서", y="매출액", hue="부서")
plt.title("부서별 총 매출액")
plt.show()
```

## 산점도 (관계)


```
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="마케팅비용", y="활성고객수", hue="부서", alpha=0.6)
plt.title("마케팅 비용과 활성고객수의 관계")
plt.show()
```

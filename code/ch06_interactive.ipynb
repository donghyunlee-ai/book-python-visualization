# 6장. 인터랙티브 차트
Plotly를 활용하여 마우스 호버와 줌인이 가능한 동적 차트를 만듭니다.


```
import pandas as pd
import plotly.express as px

df = pd.read_csv("data/tech_startup_sales.csv")
# 샘플 데이터 추출
df_sample = df.head(100)
```

## 동적 산점도 (툴팁 활용)


```
fig = px.scatter(df_sample, x="마케팅비용", y="활성고객수", color="부서", size="매출액", hover_data=["날짜"])
fig.update_layout(title="마케팅 비용 대비 활성고객수 (버블 차트)")
fig.show()
```

## 시계열 라인 차트 (Range Slider)


```
df_sales = df[df["부서"]=="영업팀"]
fig2 = px.line(df_sales, x="날짜", y="매출액", title="영업팀 2023년 매출 추이")
fig2.update_xaxes(rangeslider_visible=True)
fig2.show()
```

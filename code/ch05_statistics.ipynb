# 5장. 통계 요약과 분포
평균의 함정을 벗어나기 위한 히스토그램과 박스플롯을 배웁니다.


```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("seaborn-v0_8-whitegrid")
plt.rc("font", family="NanumBarunGothic")
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv("data/tech_startup_sales.csv")
```

## 히스토그램 (분포 파악)


```
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="매출액", hue="부서", kde=True, alpha=0.6)
plt.title("부서별 매출액 분포 (히스토그램)")
plt.show()
```

## 박스플롯 (이상치 확인)


```
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="부서", y="매출액", palette="Set2")
plt.title("부서별 매출액 분포 (박스플롯)")
plt.show()
```

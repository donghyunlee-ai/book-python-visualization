# 4장. 디자인 튜닝 공통 요소
프리미엄 디자인을 적용하는 방법을 배웁니다.


```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("seaborn-v0_8-whitegrid")
plt.rc("font", family="NanumBarunGothic")
plt.rcParams["axes.unicode_minus"] = False

df = pd.read_csv("data/tech_startup_sales.csv")
df["날짜"] = pd.to_datetime(df["날짜"])
```

```
# 프리미엄 차트 튜닝
monthly = df[df["부서"]=="영업팀"].groupby(df["날짜"].dt.to_period("M"))["매출액"].sum().reset_index()
monthly["날짜"] = monthly["날짜"].astype(str)

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(monthly["날짜"], monthly["매출액"], color="#264653", linewidth=3, marker="o")

# 주석 달기
max_idx = monthly["매출액"].idxmax()
ax.annotate("최대 매출 달성!", 
            xy=(max_idx, monthly.loc[max_idx, "매출액"]), 
            xytext=(max_idx-2, monthly.loc[max_idx, "매출액"]+1000),
            arrowprops=dict(facecolor="#e76f51", shrink=0.05), 
            weight="bold", color="#e76f51")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.set_title("영업팀 월간 매출 추이", pad=15, weight="bold")
plt.xticks(rotation=45)
plt.show()
```
